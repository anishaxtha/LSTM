```python
#  make the text into the indices
def text_to_indices(sentence, vocab):

    numerical_sentence = []

    for token in sentence:
        if token in vocab:
            numerical_sentence.append(vocab[token])
        else:
            numerical_sentence.append(vocab['<unk>'])

    return numerical_sentence
```

```python
input_numerical_sentences = []

for sentence in input_sentences:
    input_numerical_sentences.append(text_to_indices(word_tokenize(sentence.lower()),vocab))

    
```

```python
    #  making the class we get 
    class CustomDataSet(Dataset):
    
        def __init__(self, X, y):
            self.X = X 
            self.y = y
    
        def __len__(self):
            return self.X.shape[0]
    
        def __getitem__(self, idx):
            return self.X[idx], self.y[idx]

        
```

```python
dataset = CustomDataSet(X, y)
```

```python
len(dataset)
```

```python
dataset[3]   #this is just an exmaples of the datasets
```

```python
#  using the data loaders
dataloader = DataLoader(dataset, batch_size =32, shuffle = True)
```

```python
#  making of the class 
class LSTMModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, 100)   #convert the word id into vectors
        self.lstm = nn.LSTM(100,150,batch_first= True)
        self.fc = nn.Linear(150, vocab_size)

    def forward(self, x):
        embedded = self.embedding(x)
        intermediate_hidden_states, (final_hidden_state, final_cell_state) = self.lstm(embedded)
        output = self.fc(final_hidden_state.squeeze(0))
        return output
        
```

```python
model = LSTMModel(len(vocab))
```

```python
model
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
model.to(device)
```

```python
epochs = 50
learning_rate = 0.001
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = learning_rate)
```

```python
# training loop
for epoch in range(epochs):
    total_loss = 0

    for batch_x, batch_y in dataloader:
        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad()
        output = model(batch_x)
        loss = criterion(output, batch_y)
        loss.backward()
        optimizer.step()
        
        total_loss = total_loss + loss.item()

    print(f"Epoch: {epoch +1} , Loss: {total_loss:.4f}")
```

```python
def prediction(model, vocab, text):

    tokenized_text = word_tokenize(text.lower())
    numerical_text = text_to_indices(tokenized_text, vocab)

    # padding
    padded_text = torch.tensor(
        [0] * (20 - len(numerical_text)) + numerical_text,
        dtype=torch.long
    ).unsqueeze(0)

    # align device (fixes cuda:0 vs cpu error)
    padded_text = padded_text.to(next(model.parameters()).device)


    # send to model
    model.eval()
    
    with torch.no_grad():
        output = model(padded_text)

    # predicted index
    _, index = torch.max(output, dim=1)
    predicted_index = index.item()

    #  safer index → word mapping
    index_to_word = {idx: word for word, idx in vocab.items()}
    predicted_word = index_to_word.get(predicted_index, "<UNK>")

    # print("Predicted word:", predicted_word)

    # merge with input text
    final_text = text + " " + predicted_word
    return final_text
```

```python
num_tokens = 10
input_text = "Classes are conducted according to a"

for i in range(num_tokens):
    output_text = prediction(model, vocab, input_text)
    print(output_text)

    if output_text.strip().endswith("."):
        break

    input_text = output_text

```
